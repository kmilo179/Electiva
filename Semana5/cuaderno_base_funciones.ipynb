## Parte 1 — Qué es una función

Una **función** es un procedimiento con nombre que Python guarda para usar cuantas veces se
necesite, sin volver a escribirlo. Se define con `def`, un nombre, paréntesis con los
**parámetros** (los datos que la función recibe) y dos puntos; el cuerpo va indentado; `return`
entrega el resultado de vuelta a quien la llamó.

Empiece con una función mínima, sin relación todavía con datos sucios, solo para ver la anatomía:

````python
# 'def' define la funcion. 'x' es el PARAMETRO: un nombre temporal para el valor
#  de entrada que se suministra pasen al llamarla.
# 'return' entrega el resultado; sin return, la funcion no devuelve nada util 
# (devuelve None).
def duplicar(x):
    resultado = x * 2
    return resultado

# Llamar la funcion: Python reemplaza 'x' por el valor que le pasamos entre 
# parentesis.
duplicar(5)
````

Prediga antes de correr qué devuelve cada llamado. Note que **el mismo código** de la función se ejecuta con datos distintos cada vez — eso es justamente lo que evita copiar y pegar:

_Duplicar va devolver el valor en parentesis sumado dos veces en este caso un entero 3.5 pasará a 7.0_

_El segundo duplicar devolvera los datos sumados dos veces, en este caso esta sumando dos Strings por lo tanto '16' dará '1616'_

````python
duplicar(3.5)
````

````python
duplicar(bikers_crudo[0])  # OJO: bikers_crudo[0] es el STRING '16', no el numero 16
````

---
### 🅢 Punto de encuentro 1 — anatomía de una función *(pausa guiada)*

- **`def nombre(parametros):`** — define la función una sola vez; el cuerpo (indentado) es el procedimiento que antes copiaba y pegaba.
- **Parámetro** — un nombre temporal (aquí `x`) que representa lo que sea que le pasen al llamar la función. Cambia en cada llamado; el código de adentro no cambia.
- **`return`** — entrega el resultado a quien llamó la función, para poder guardarlo en una variable o usarlo en otra parte.
- **Llamar la función** (`duplicar(5)`) — ejecuta el cuerpo con `x = 5` y devuelve `10`.

## Parte 2 — Encapsule la corrección en una función

Complete la función `reconciliar_columna`. El **contenido de la función** (lo que promete hacer, sin importar qué columna reciba) ya está escrito en el `docstring` y reforzado con un `assert` al final —una línea que **detiene el programa con un error si el resultado no cumple lo prometido**. Su trabajo es escribir el cuerpo, reusando la lógica que ya escribió a mano en la semana 3.

````python
def reconciliar_columna(columna_cruda, tipo_destino):
    """
    Recibe una columna sucia (lista de strings) y el tipo al que debe convertirse
    (la propia funcion int o float de Python, pasada como valor).
    Devuelve una lista nueva, mas corta o igual, ya convertida a tipo_destino:
      - descarta los valores 'n/d' (no los convierte en 0),
      - repara la coma decimal antes de convertir (.replace(',', '.')),
      - convierte cada valor restante con tipo_destino(valor).
    """
    columna_limpia = []
    for valor in columna_cruda:
        if valor == 'n/d':
            continue
        else:
            #valor.replace(',', '.')
            columna_limpia.append(tipo_destino(valor.replace(',', '.')))
        
        # TODO 2 -- Ahora usted: escriba aqui el cuerpo del for.
        #   Reutilice la logica de la Parte 2 de la semana 3:
        #   1) si valor == 'n/d', descartelo (continue) sin agregarlo a columna_limpia
        #   2) si no, repare la coma con .replace(',', '.')
        #   3) conviertalo con tipo_destino(...) y agreguelo a columna_limpia con .append(...)
        #pass  # <-- reemplace esta linea

    # El contenido de la funcion, verificado automaticamente cada vez que se llama:
    assert len(columna_limpia) <= len(columna_cruda), 'no puede sobrar mas datos limpios que crudos'
    assert all(isinstance(v, tipo_destino) for v in columna_limpia), 'todo debe quedar en el tipo pedido'
    assert len(columna_limpia) > 0, 'La lista columna_limpia debe tener más de 0 valores'
    return columna_limpia
````

**Qué hace cada comando:**

| Comando | Qué hace |
|---|---|
| `git init` | Crea el repositorio (una carpeta oculta `.git/`) — se hace **una sola vez** por proyecto |
| `git add .` | Marca los archivos actuales como listos para el próximo commit ("preparación") |
| `git commit -m "..."` | Guarda una **fotografía** permanente de los archivos preparados, con un mensaje |
| `git log` | Muestra el historial de commits — quién, cuándo, y con qué mensaje |
