### 🅢 Punto de encuentro 1 — lo que acaba de descubrir *(pausa guiada)*

Póngale nombre a lo que vivió:

- **Kernel** = la *memoria viva* del cuaderno. Guarda el valor de cada variable mientras la sesión está abierta.
- **Celda** = una instrucción que se *envía* al kernel. El kernel la ejecuta con base en la información con que se cuenta **en ese instante**.
- **`[n]`** (el número entre corchetes a la izquierda de cada celda) = es el **orden real** en que se ejecutaron las celdas, no el orden en que están escritas.
- Un **"resultado raro"** casi siempre es **estado viejo**: el kernel recuerda algo de una corrida anterior.

La herramienta que corta ese enredo es **`Kernel → Restart & Run All`**: borra la memoria y corre todo desde cero, en orden. Es la prueba de honestidad de un cuaderno.
