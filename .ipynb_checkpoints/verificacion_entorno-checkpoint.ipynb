## 6. `[ETAPA ?]` — Limpiar, decidiendo

Cada línea de abajo es **una decisión**. No hay una única forma correcta; pero sí hay que poder justificar la decisión que se tome.

```python
bikers = (df['bikers']
          .astype(str)
          .str.strip()
          .replace('n/d', np.nan))          # decisión: 'n/d' = dato faltante
bikers = pd.to_numeric(bikers, errors='coerce')  # lo que no sea número -> NaN

temp = (df['temp'].astype(str).str.strip()
        .str.replace(',', '.', regex=False)) # decisión: la coma es separador decimal
temp = pd.to_numeric(temp, errors='coerce')

limpio = df.assign(bikers=bikers, temp=temp)
limpio = limpio.drop_duplicates()            # decisión: quitar filas repetidas
limpio = limpio[limpio['bikers'] > 0]        # decisión: alquileres <= 0 es imposible

print('Filas antes:', len(df), '-> después de limpiar:', len(limpio))
```

```python
df['bikers'].replace('n/d', np.nan).astype(float).sort_values()
```

## 7. `[ETAPA ?]` — Ahora sí, a explorar

Con los datos limpios, ya podemos extraer datos útiles del resumen.

```python
print('Alquileres promedio por hora:', round(limpio['bikers'].mean(), 1))
print('Alquileres totales (muestra):', int(limpio['bikers'].sum()))
print()
# Promedio por clima (normalizando la categoría):
limpio = limpio.assign(weathersit=limpio['weathersit'].str.strip().str.lower())
limpio.groupby('weathersit')['bikers'].mean().round(1)
```
