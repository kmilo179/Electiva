## Parte 2 — Tuplas: el contrapunto que no cambia

Bogotá tiene una latitud y una longitud. Ese par **no debería poder cambiar** una vez definido —no
es una lista de cosas que crece, es un dato fijo con dos partes. Para eso existe la **tupla**: se
escribe casi igual que una lista, pero con paréntesis `()` en vez de corchetes.


```python
coordenada_bogota = (4.6097, -74.0817)  # (latitud, longitud)
coordenada_bogota
```

```python
coordenada_bogota[0]
```

Prediga qué pasa si intenta **cambiar** un elemento de la tupla, igual que cambiaría uno de una
lista (`mpg[0] = 99.0` funcionaría sin problema). **No borre el error si aparece**:


```python
coordenada_bogota[0] = 5.0
```

## Parte 3 — Diccionarios: cuando cada valor tiene un nombre

Una lista como `mpg` guarda números sueltos, pero cada auto real trae **varios datos juntos**:
nombre, rendimiento, cilindros, caballos de fuerza... Guardarlos como `auto_nombre_1`,
`auto_mpg_1`, `auto_cilindros_1` sería tan incómodo como los `mpg_1`, `mpg_2` del principio.

Un **diccionario** guarda pares **nombre → valor**, con llaves `{}`:


```python
# Un auto real del dataset representado como diccionario.
auto1 = {
    'name': 'chevrolet chevelle malibu',
    'mpg': 18.0,
    'cylinders': 8,
    'horsepower': 130,
    'weight': 3504,
    'year': 70,
}
auto1
```

Prediga qué devuelve cada acceso. Ya no se pide por **posición** (`mpg[0]`) sino por **nombre**:


```python
auto1['mpg']
```

```python
auto1['name']
```

### 🅢 Punto de encuentro 1 — nombrar las tres estructuras *(pausa guiada)*

- **Lista** `[ ]` — varios valores, en orden, **se pueden modificar** (`.append()` agrega). Se
  accede por **posición**: `mpg[0]`.
- **Tupla** `( )` — igual que una lista pero **fija**: no se puede cambiar después de creada. Sirve
  para datos que tienen partes que no deberían moverse (una coordenada, una fecha).
- **Diccionario** `{ }` — pares **nombre → valor**. Se accede por **nombre**, no por posición:
  `auto1['mpg']`. Una **lista de diccionarios** (`autos`) es un mini-dataset: cada diccionario es un
  registro (una fila), cada nombre es una columna.

El nombre llegó **después** de que ya los usó. Ahora, con `autos` armado, hagamos preguntas sobre
el conjunto completo.

## Parte 4 — Consultas simples con `for`

Recorrer una lista de diccionarios uno por uno, con `for`, es como pasar hoja por hoja una lista de
fichas. Prediga qué imprime esta celda **antes de correrla**:


```python
for auto in autos:
    print(auto['name'], '->', auto['mpg'], 'mpg')
```

Ahora una consulta que **acumula** un total mientras recorre la lista — la misma idea que va a necesitar después para reconciliar una columna completa de datos sucios:


```python
total_mpg = 0
for auto in autos:
    total_mpg = total_mpg + auto['mpg']

print('mpg total de la muestra:', total_mpg)
print('mpg promedio:', total_mpg / len(autos))
```

## Parte 5 — Conjuntos: ¿Cuántos valores *distintos* hay?

Última estructura. `autos` trae dos valores de `cylinders` que se repiten varias veces (`8` y `4`).
Para saber cuántos valores **distintos** aparecen, sin contarlos a mano ni repetir ninguno dos
veces, existe el **conjunto** (`set`).


```python
cilindros = []
for auto in autos:
    cilindros.append(auto['cylinders'])

print('todos los valores, con repeticion:', cilindros)
print('valores distintos:', set(cilindros))
```
