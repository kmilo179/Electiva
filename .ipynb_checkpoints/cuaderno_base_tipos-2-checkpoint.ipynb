### 🅢 Punto de encuentro 1 — los cuatro tipos *(pausa guiada)*

- **`int`** — número entero: `7`, `-3`, `0`.
- **`float`** — número con decimales: `7.0`, `3.5`, `0.24`. (Por eso `10 / 3` da `3.333...`: la división `/` siempre da `float`.)
- **`str`** — texto (*string*): `'7'`, `'hola'`. Lleva comillas. **`'7'` no es el número 7.**
- **`bool`** — verdadero/falso: el resultado de una pregunta sí/no como `5 > 3`.

El nombre llegó **después** de que ya los vio. Y el `+`: con números suma, con `str` y con listas **pega** (concatena).
