### 🅢 Punto de encuentro 1 — nombrar la tabla *(pausa guiada)*

- **`DataFrame`** — la tabla completa: filas y columnas, con nombres y tipos. Es el sucesor natural de la **lista de diccionarios** (`autos`); puede incluir miles de filas.
- **Fila** — un registro completo (un auto). Antes era `autos[i]`, un diccionario.
- **Columna** — una variable (`mpg`, `weight`, ...). Antes era la llave de cada diccionario.
- **`Series`** — una sola columna del DataFrame, sacada aparte. Se ve como una lista con un
  nombre y un tipo pegados.
- **`pd.read_csv(...)`** — reemplaza escribir la lista de diccionarios a mano: lee el archivo y
  arma el DataFrame completo.

## Parte 4 — De la posición en la lista a la posición en la tabla

En `autos`, `autos[0]` permitìa obtener el primer diccionario **por posición** — igual que `mpg[0]` en la
semana 3 con una lista de números. Un DataFrame también puede ser indexado por fila, con `.iloc[]` (*integer location*).

_(reemplace esta línea)_

```python
print('Por diccionario:', autos[0])
print()
print('Por DataFrame (.iloc[0]):')
print(df.iloc[0])
```

**`# TODO 5`** — Compare las dos salidas de la celda anterior. En sus palabras: ¿En qué se parece `df.iloc[0]` a `autos[0]` y qué trae de más? _(reemplace esta línea)_

Ahora, una verificación más: si el `DataFrame` realmente contiene los 392 registros del conjunto de datos original, entonces las **mismas ocho filas** que usted escribió a mano en la Parte 0 deben también estar ahí, en las posiciones `0, 1, 2, 3, 4, 6, 58` y `59` (no son consecutivas).

Prediga: si suma `mpg` de esas ocho filas del DataFrame con `.sum()` ¿Debería dar el mismo
`total_mpg` que ya calculó a mano en la Parte 0? _(reemplace esta línea)_

```python
posiciones_de_la_muestra = [0, 1, 2, 3, 4, 6, 58, 59]
total_mpg_pandas = df.iloc[posiciones_de_la_muestra]['mpg'].sum()

print('total mpg con pandas (mismas 8 filas):', total_mpg_pandas)
print('total mpg a mano, Parte 0:', total_mpg)

assert total_mpg_pandas == total_mpg, 'las 8 filas deberian sumar exactamente lo mismo'
print('Checkpoint superado: el DataFrame trae, sin alterar, los mismos datos que reconcilio a mano.')
```

## Parte 5 — ⭐ Filtrar sin escribir un ciclo `for`

En el _notebook_ de la semana 3 (`# TODO 6`) usted escribió un `for` con un `if` adentro para imprimir solo los autos con `mpg > 16`. Sobre los 8 autos de la muestra, la respuesta correcta era **5** autos. pandas puede hacer la misma pregunta sin ningún `for`, con una **condición booleana** directamente sobre la columna — la misma idea de máscara que usó con `numpy` en la semana 5, pero ahora sobre una tabla completa.

Prediga: si aplica `df.iloc[posiciones_de_la_muestra]['mpg'] > 16` sobre los mismos 8 autos,
¿Debería contar también 5 valores `True`? _(reemplace esta línea)_

```python
mascara_muestra = df.iloc[posiciones_de_la_muestra]['mpg'] > 16
print(mascara_muestra)
print('cuantos True:', mascara_muestra.sum())  # sum() de booleanos cuenta los True

assert mascara_muestra.sum() == 5, 'deberia coincidir con el conteo manual de la semana 3'
print('Checkpoint superado: coincide con el conteo manual de la semana 3.')
```

## Parte 6 — Predecir vs. inferir

### 🅢 Punto de encuentro 2 — predicción, inferencia y el vocabulario de la Unidad 2 *(pausa guiada)*

- **Predicción** — usar datos que sí conoce para adivinar un valor que no conoce (¿Cuánto consumirá este auto nuevo?).
- **Inferencia** — entender **por qué** dos variables están relacionadas (¿El peso realmente afecta el consumo, o es casualidad en esta muestra?). Distinta pregunta, misma tabla.
- **Variable objetivo** — la columna que quiere predecir o explicar (`mpg`, en el ejemplo de hoy).
- **Predictores** — las columnas que usaría como pista para esa predicción.
