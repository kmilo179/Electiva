### 🅢 Punto de encuentro 1 — casting y sus dos formas de fallar *(pausa guiada)*

- **`int(texto)`, `float(texto)`, `str(valor)`** convierten **a propósito**, cuando el texto
  *tiene forma* de número.
- **`'n/d'` no tiene forma de número en ningún idioma**: no hay conversión posible, es un dato
  **faltante** de verdad. La decisión correcta no es forzar un `0` —es **descartar la fila** o
  marcarla aparte.
- **`'0,24'` sí es un número, escrito con la convención decimal equivocada** para Python (que
  espera punto, no coma). Se repara **antes** del `float()`, cambiando la coma por un punto con
  `.replace(',', '.')`.

Dos problemas que se ven igual de "rotos" al fallar, pero se reparan distinto. Vamos a reconciliar
la columna completa.

## Parte 2 — Reconcilie `bikers` (la columna de la semana 1)

Esta es, sin cambiar ni un carácter, la columna `bikers` de `bikeshare_sample.csv` (semana 1): las
bicicletas alquiladas por hora, tal como llegaron del archivo —como **texto**, no como número
(así es como Python lee *cualquier* columna de un CSV sin ayuda).


```python
bikers_crudo = ['16', '40', 'n/d', '1', '36', '34', '28', '9', '8', 'n/d', '57', '-5', '82', '92', '51', '34']
len(bikers_crudo)
```

Prediga qué imprime esto —**¿lo reconoce?**— antes de correrlo:

_Va contar cuantos valores hay en la lista_

```python
# Sumar con + una lista de STRINGS no suma numeros: concatena texto, uno detras de otro.
total_ingenuo = ''
for valor in bikers_crudo:
    total_ingenuo = total_ingenuo + valor

print(total_ingenuo)
```

**`# TODO 3`** — Es el mismo total-basura de la semana 1
(`'1640n/d136342898n/d57-582925134'`). En la semana 1 lo vio salir de una línea de pandas que no
explicamos. Ahora que sabe por qué pasa: en una frase, ¿qué hace exactamente el `+` cuando ambos
lados son `str`? _Lo que hace es coger ese string y lo fusiona con el otro string, por ejemplo si tenemos dos variables que es v1 = "hola " y otra que sea v2 = "mundo", al hacer esto v3 = v1 + v2, la variable 3 será: "hola mundo"_

Repárelo. La celda de abajo recorre `bikers_crudo` y decide, **valor por valor**, qué hacer:


```python
bikers_limpio = []
for valor in bikers_crudo:
    if valor == 'n/d':
        continue  # decision: 'n/d' se DESCARTA, no se convierte en 0
    bikers_limpio.append(int(valor))

print('valores originales:', len(bikers_crudo))
print('valores utilizables:', len(bikers_limpio))
bikers_limpio
```

**`# TODO 4`** — ¿Por qué **no** conviene reemplazar `'n/d'` por `0` en vez de descartar la fila?
Piense en lo que significaría un `0` en esta columna: *cero bicicletas alquiladas esa hora* es un
dato real y posible. ¿Qué pasaría con el promedio si mete ceros donde en realidad **no se sabe**
cuántas se alquilaron? _porque el 0 tambien puede ser un valor de la lista, por eso es mejor utilizar directamente una cadena que se pueda descartar en la lista o sistema, en este caso ("n/d"). Si en promedio se metiera cetos donde en realidad no se sabe, ese '0' como un valor más de la lista y al realizar operaciones importantes como el promedio afectaría el valor final_


```python
total_bikers = 0
for valor in bikers_limpio:
    total_bikers = total_bikers + valor

promedio_bikers = total_bikers / len(bikers_limpio)  # OJO: se divide por len(bikers_limpio), NO por len(bikers_crudo)
print('total de bicicletas alquiladas (filas utilizables):', total_bikers)
print('promedio por hora:', round(promedio_bikers, 1))

# Atajo equivalente al for-loop de arriba, con la funcion incorporada sum():
print('mismo total con sum():', sum(bikers_limpio))
```

> **Nota aparte (no se repara hoy):** hay un `-5` en los datos originales. Que Python pueda sumar
> `-5` no significa que tenga sentido: *-5 bicicletas alquiladas* no es un tipo equivocado, es un
> **valor** que no tiene sentido físico. Ese problema se ataca en la semana 8 (análisis
> exploratorio); hoy el trabajo es solo de **tipos**.


---
### 🅢 Punto de encuentro 2 — el puente de la honestidad *(pausa guiada)*

`promedio_bikers` se dividió por `len(bikers_limpio)` (**14** valores utilizables), no por
`len(bikers_crudo)` (**16** valores totales). Si hubiera dividido por 16, el promedio saldría más
bajo de lo real —como si los dos `'n/d'` contaran como *cero bicicletas* en vez de *no sabemos*.

Esta es la **invariante de honestidad** del curso, aplicada por primera vez: una decisión de
limpieza (qué hacer con lo faltante) **cambia el resultado**, y hay que poder justificarla. No hay
una única respuesta correcta —descartar la fila es una decisión razonable, pero debe **quedar
explícita**, no escondida en el código.
